- Process
    1. 이탈 고객 예측 classifier 모델 생성
    2. 이 모델을 기반으로 전체 데이터 기준으로 적용 후 product_proba 값을 도출
    3. product_proba 값을 기반으로 하는 고객 군집  수행
    	- 이탈 여부에 기반한 Segmentation (e.g., RFM, KMeans, GMM)
    	- 고객을 이탈/유지 여부 + 특성 기반으로 군집화하여, 타겟 마케팅 그룹 도출
    	- 고객별 위험 등급 부여 (ex: High, Medium, Low)
    4. 고객별 위험 등급별 이탈/유지 에 영향을 주는 요소를 파악
    	- 인구 통계학적 요소
    	- 신용카드 제공 업체와의 관계 관련 요소 
    		- low 고객의 특성에서는 이런 부분이 특성을 갖고 있다
    		- medium 고객은 이런 특성을 갖고 있다
    		- high 고객은 이런 특성을 갖고 있다
    	- 이를 통해 이탈 high 고객은 이런 특성에서 중요도 높게 신경을 써야겠다 라는 인사이트 도출 
    5. 도출된 특성을 가장 효율적으로 관리하기 위해서는 어떤 특성을 가진 고객군 대상으로 타겟팅하면 좋을까
    	- 도출된 특성 기준으로 인구통계학적 특성에 기반한 분포 분석을 통해 특성별 가진 세그먼트를 도출
    		- 여기서도 군집 모델을 사용해야 할지는 확인이 필요합니다.
    			- 가장 효율적인 이탈 방지 위해서는 x,y 특성을 가진 고객군 대상으로 집중 매니지먼트 수행 제안 으로 결론

- 예상 최종 제안 
    - 이탈 예측 모델에 따르면, 총거래금액, 평균 구매가능 한도, 비활성 기간이 이탈에 가장 크게 기여합니다.
    - 특히 36개월 이상 보유 고객 중에서도 거래 없는 고한도 고객은 이탈률이 높아, 우선 관리 타겟으로 분류됩니다.
    - 클러스터 분석 결과, RFM 점수가 낮고 평균 이용 한도가 높은 군집은 전체 이탈의 40%를 차지하며, 이 군집에 대한 별도 정책 수립이 필요합니다.
    - 성별·연령 분석에 따르면, 50대 남성의 이탈률이 가장 높고, 20대 여성은 가장 낮은 경향을 보여 연령·성별 기반 맞춤 관리 방안이 효과적입니다.
    - 고객에게 부여한 Churn Risk Score를 기반으로, High Risk 고객 15%에 대해 인센티브 제공 또는 인입콜 전략을 제안하며,
    동시에 rule-based로 “거래 없음 + 한도 높음” 고객을 High로 자동 지정하는 정책을 도입할 수 있습니다.

# 9. LightGBM - Feature Selection - Feature Importance & Permutation Importance
- LightGBM 학습 (Scaling 전 전체 feature 사용)
- Feature Importance 추출 (importance_type='gain' 기준)
- 중요도 누적 상위 5개 feature 선정 (80%)
- 5개 feature로 다시 LightGBM 재학습

````python
# 9. LightGBM - Feature Importance

# x columns, target columns split
x = df_preprocessed_for_modeling_without_scaling.drop(['Attrition_Flag'], axis = 1)
y = df_preprocessed_for_modeling_without_scaling['Attrition_Flag']

x_train, x_val, y_train, y_val = train_test_split(x, y, test_size = 0.2, random_state = 42)

print(len(x_train), len(y_train), len(x_val), len(y_val))
print()

# LGBMClassifier
model = LGBMClassifier(random_state = 42)
model.fit(x_train, y_train)

# # gain 기반 중요도 추출 : 
# importances_gain = model.booster_.feature_importance(importance_type='gain')
# feature_importances = pd.Series(importances_gain, index=x_train.columns)

# 9. LightGBM - Permutation Importance
from sklearn.inspection import permutation_importance

# Permutation Importance
result = permutation_importance(model, x_val, y_val, scoring='roc_auc', n_repeats=10, random_state=42)

# 정리
import pandas as pd
import matplotlib.pyplot as plt

importance_df = pd.DataFrame({
    'feature': x_val.columns,
    'gain_importance': model.booster_.feature_importance(importance_type='gain'),
    'perm_importance': result.importances_mean
}).sort_values(by='perm_importance', ascending=False)

# cumulative 추가
importance_df['gain_importance_cumulative'] = importance_df['gain_importance'].cumsum() / importance_df['gain_importance'].sum()
importance_df['perm_importance_cumulative'] = importance_df['perm_importance'].cumsum() / importance_df['perm_importance'].sum()

importance_df.reset_index(drop = True)
````

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 데이터 준비
top_n = 10
plot_df = importance_df.head(top_n).copy()

# Melt for Seaborn
plot_melt = plot_df.melt(id_vars='feature', 
                         value_vars=['gain_importance', 'perm_importance'],
                         var_name='importance_type', value_name='score')

plt.figure(figsize=(10, 5))
plt.plot(plot_df['feature'], plot_df['gain_importance_cumulative'], marker='o', label='Cumulative Gain')
plt.plot(plot_df['feature'], plot_df['perm_importance_cumulative'], marker='s', label='Cumulative Permutation')

plt.title('Cumulative Feature Importance')
plt.ylabel('Cumulative Proportion')
plt.xticks(rotation=45, ha='right')
plt.ylim(0, 1.05)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
````

````python
def plot_mean_median_mode(df, col): 
    fig, axes = plt.subplots(1, 3, figsize=(14, 4))

    # 값 계산
    mean_val = df[col].mean()
    median_val = df[col].median()
    mode_val = df[col].mode().iloc[0]  # 최빈값 여러 개일 경우 첫 번째 사용

    # 히스토그램
    sns.histplot(df[col], ax=axes[0], kde=True)
    axes[0].axvline(mean_val, color='red', linestyle='--', label='Mean')
    axes[0].axvline(median_val, color='green', linestyle='-.', label='Median')
    axes[0].axvline(mode_val, color='blue', linestyle=':', label='Mode')
    axes[0].legend()
    axes[0].set_title(f"Histogram of {col}")

    # 박스플롯
    sns.boxplot(x=df[col], ax=axes[1])
    axes[1].axvline(mean_val, color='red', linestyle='--', label='Mean')
    axes[1].axvline(median_val, color='green', linestyle='-.', label='Median')
    axes[1].axvline(mode_val, color='blue', linestyle=':', label='Mode')
    axes[1].legend()
    axes[1].set_title(f"Boxplot of {col}")

    # 바이올린플롯
    sns.violinplot(x=df[col], ax=axes[2], fill=False)
    axes[2].axvline(mean_val, color='red', linestyle='--', label='Mean')
    axes[2].axvline(median_val, color='green', linestyle='-.', label='Median')
    axes[2].axvline(mode_val, color='blue', linestyle=':', label='Mode')
    axes[2].legend()
    axes[2].set_title(f"Violinplot of {col}")

    plt.tight_layout()
    plt.show()
    
    # 
    print(tmp[col].describe())
````

````python
# *_value: 해당 feature의 클러스터별 기술통계값 (benchmark 기준 방식과 동일하게 계산)

# *_diff: benchmark 값과의 차이
# (예: 클러스터의 평균 - benchmark 평균)

import pandas as pd

tmp = x_existing_with_proba_Attrition_cluster_fi

# benchmark 정의
benchmark = {
    'Total_Trans_Ct': x_existing_with_proba_Attrition_cluster_fi[x_existing_with_proba_Attrition_cluster_fi['cluster_fi'] == 0]['Total_Trans_Ct'].mode().iloc[0],  # mode
    'Total_Trans_Amt': x_existing_with_proba_Attrition_cluster_fi[x_existing_with_proba_Attrition_cluster_fi['cluster_fi'] == 0]['Total_Trans_Amt'].mean(),  # mean
    'Total_Revolving_Bal': x_existing_with_proba_Attrition_cluster_fi[x_existing_with_proba_Attrition_cluster_fi['cluster_fi'] == 0]['Total_Revolving_Bal'].median(),  # median
    'Total_Relationship_Count': x_existing_with_proba_Attrition_cluster_fi[x_existing_with_proba_Attrition_cluster_fi['cluster_fi'] == 0]['Total_Relationship_Count'].median(),  # median
    'Total_Amt_Chng_Q4_Q1': x_existing_with_proba_Attrition_cluster_fi[x_existing_with_proba_Attrition_cluster_fi['cluster_fi'] == 0]['Total_Amt_Chng_Q4_Q1'].mean()  # mean
}

# 기준 방식 정의
stat_method = {
    'Total_Trans_Ct': 'mode',
    'Total_Trans_Amt': 'mean',
    'Total_Revolving_Bal': 'median',
    'Total_Relationship_Count': 'median',
    'Total_Amt_Chng_Q4_Q1': 'mean'
}

# 클러스터별 비교
features = list(benchmark.keys())
cluster_col = 'cluster_fi'
cluster_diff_result = []

for cluster_id, group in x_existing_with_proba_Attrition_cluster_fi.groupby(cluster_col):
    row_result = {'cluster': cluster_id}
    
    for feature in features:
        method = stat_method[feature]
        if method == 'mean':
            cluster_value = group[feature].mean()
        elif method == 'median':
            cluster_value = group[feature].median()
        elif method == 'mode':
            cluster_value = group[feature].mode().iloc[0] if not group[feature].mode().empty else None
        else:
            cluster_value = None
        
        benchmark_value = benchmark[feature]
        row_result[f'{feature}_value'] = cluster_value
        row_result[f'{feature}_diff'] = cluster_value - benchmark_value if cluster_value is not None else None

    cluster_diff_result.append(row_result)

# 결과 출력
diff_df = pd.DataFrame(cluster_diff_result)

````
