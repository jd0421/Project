````python
# for numerical data
def top5_proportion(target) :
    # top 5 선별, 전체 대비 비중
    tmp = df_final.groupby([target])['정기예금 가입여부'].agg('count').reset_index().sort_values('정기예금 가입여부', ascending = False)
    tmp['proportion'] =  tmp['정기예금 가입여부'] / tmp['정기예금 가입여부'].sum()
    tmp = tmp.head(5)
    display(tmp)

    # 데이터프레임 준비
    tmp = df_final.groupby([target, '정기예금 가입여부']).size().reset_index().sort_values('정기예금 가입여부', ascending = False)
    tmp.columns = [target, '정기예금 가입여부', 'count']
    tmp['proportion'] = tmp['count'] / tmp['count'].sum()
    tmp = tmp.sort_values([target, '정기예금 가입여부'], ascending = [False, False])
    tmp = tmp.reset_index(drop = True)

    # 기존 순서 추출
    original_order = tmp[target].drop_duplicates().tolist()

    #  피벗 테이블 생성
    pivot_df = tmp.pivot(index=target, columns='정기예금 가입여부', values='count').fillna(0)
    pivot_df.columns = ['False', 'True']
    pivot_df = pivot_df[['False', 'True']]  # 순서 재정렬

    # 기존 순서대로 재정렬
    pivot_df = pivot_df.loc[original_order]

    # 비율 계산 (row-wise)
    proportion_df = pivot_df.div(pivot_df.sum(axis=1), axis=0)

    # 시각화
    ax = pivot_df.plot(
        kind='bar',
        stacked=True,
        figsize=(12, 3),
        color=['#dd8452', '#4c72b0']
    )

    # 비율 라벨 표시
    for idx, (false_val, true_val) in enumerate(zip(pivot_df['False'], pivot_df['True'])):
        total = false_val + true_val
        if false_val > 0:
            ax.text(
                idx, false_val / 2, f"{proportion_df.iloc[idx]['False']:.0%}",
                ha='center', va='center', fontsize=9, color='white'
            )
        if true_val > 0:
            ax.text(
                idx, false_val + (true_val / 2), f"{proportion_df.iloc[idx]['True']:.0%}",
                ha='center', va='center', fontsize=9, color='white'
            )

    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()
````
