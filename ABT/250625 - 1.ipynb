###  실험/대조군 간 퍼널(Funnel) 단축 효과 분석
```
- 서비스별 진입 경로에 따라 탐색 깊이(click_to_entry_depth)가 어떻게 달라졌는가를 비교하여,
노출 구조 변경(H1)이 유저의 진입 퍼널을 얼마나 단축시켰는지 정량적으로 평가하겠습니다.
```

````python
filtered_df = df[
    df['service_name'].isin(['대출 한도', '투자 상품', '내 카드 한도']) &
    df['entry_point'].isin(['홈', '더보기', '검색', '햄버거'])
]

# 그룹: 실험군/대조군, 진입 경로, 서비스명
funnel_result = filtered_df.groupby(['experiment_group', 'entry_point', 'service_name'])[
    'click_to_entry_depth'
].mean().reset_index()

# Pivot해서 실험군/대조군 비교
pivot_df = funnel_result.pivot(index=['entry_point', 'service_name'], columns='experiment_group', values='click_to_entry_depth').reset_index()
pivot_df['depth_gap'] = pivot_df['H0'] - pivot_df['H1']

# 불필요 컬럼 experiment_group 제외 및 entry_point == '홈' 항목만 조회를 통해서 홈 화면에서 '중개서비스' 로 바로 진입할 수 있도록 구현한 h1 환경에서의 성과를 집중하여 보겠습니다.
# pivot_df.pop('experiment_group')
pivot_df[pivot_df['entry_point'] == '홈'].style.background_gradient(cmap='coolwarm')

# comment
# 실험군(H1)에서는 '홈 화면 → 중개서비스' 진입 시 평균 클릭 깊이가 1.27~1.47단계 단축되었습니다.
# 이는 홈 화면 구조 변경이 사용자 진입 경로 간소화에 실질적인 기여를 했음을 시사합니다.
````

### 1차 요약
```
Summary
- 실험군(H1)은 대조군(H0) 대비
    - 클릭률이 약 0.62%p 높아졌습니다.
    - 탐색 깊이가 1.37단계 감소, 페이지 뷰 수는 1.35회 감소했으나, 세션 체류시간은 약 3.5초 증가하였습니다.
    - 또한 이탈률도 0.6%p 개선되며, 노출 구조 개선(H1)이 사용자의 집중된 탐색 흐름을 유도한 것으로 해석할 수 있습니다.
    - 실험군(H1)에서는 '홈 화면 → 중개서비스' 진입 시 평균 클릭 깊이가 1.27~1.47단계 단축되었습니다.
    - 이는 홈 화면 구조 변경이 사용자 진입 경로 간소화에 실질적인 기여를 했음을 시사합니다.

```


````python
# 전체 KPI 평균 값 (비율 또는 평균)
overall_metrics = {
    'click_rate': df['click_service'].mean(),
    'click_to_entry_depth': df['click_to_entry_depth'].mean(),
    'click_count': df['click_count'].mean(),
    'page_view_count': df['page_view_count'].mean(),
    'session_duration_sec': df['session_duration_sec'].mean(),
    'dropout_rate': df['dropout_flag'].mean(),
    'silent_drop_rate': df[(df['click_service'] == 0) & (df['dropout_flag'] == 1)].shape[0] / len(df)
}

# 실험 그룹별 KPI 값
group_metrics = df.groupby('experiment_group').agg({
    'click_service': 'mean',
    'click_to_entry_depth': 'mean',
    'click_count': 'mean',
    'page_view_count': 'mean',
    'session_duration_sec': 'mean',
    'dropout_flag': 'mean'
})

# 진입 실패 + 클릭 없음 비율 (group별)
group_total = df.groupby('experiment_group').size()
group_silent_drop_counts = df[(df['click_service'] == 0) & (df['dropout_flag'] == 1)].groupby('experiment_group').size()
group_silent_drop_rate = group_silent_drop_counts / group_total

# group_metrics에 silent_drop_rate 컬럼 추가
group_metrics['silent_drop_rate'] = group_silent_drop_rate

# 'avg' row로 전체 평균 값 추가
overall_df = pd.DataFrame(overall_metrics, index=['avg'])

# 컬럼명 수정
overall_df.columns = ['click_service', 'click_to_entry_depth', 'click_count', 'page_view_count', 'session_duration_sec', 'dropout_flag', 'silent_drop_rate']

full_metrics_df = pd.concat([group_metrics, overall_df])

# 결과 출력
full_metrics_df.style.background_gradient(cmap='coolwarm')
````

### 실험/대조군 유입 경로 시각화 : Sankey chart

````python
import pandas as pd
import plotly.graph_objects as go

df_sankey = df[['experiment_group', 'entry_point', 'service_name', 'dropout_flag']].dropna()
df_sankey['dropout_flag'] = df_sankey['dropout_flag'].map({1: '이탈', 0: '전환'})

for group in df_sankey['experiment_group'].unique():
    subset = df_sankey[df_sankey['experiment_group'] == group]

    group_flows = pd.concat([
        subset.groupby(['entry_point', 'service_name']).size().rename("count1").reset_index(),
        subset.groupby(['service_name', 'dropout_flag']).size().rename("count2").reset_index()
    ])
    labels = pd.unique(group_flows[['entry_point', 'service_name', 'dropout_flag']].values.ravel())
    label_map = {label: i for i, label in enumerate(labels)}

    link1 = subset.groupby(['entry_point', 'service_name']).size().reset_index(name='count')
    link1['source'] = link1['entry_point'].map(label_map)
    link1['target'] = link1['service_name'].map(label_map)

    link2 = subset.groupby(['service_name', 'dropout_flag']).size().reset_index(name='count')
    link2['source'] = link2['service_name'].map(label_map)
    link2['target'] = link2['dropout_flag'].map(label_map)

    links_df = pd.concat([link1, link2])

    sankey = go.Figure(data=[go.Sankey(
        arrangement="snap",
        node=dict(
            pad=15,
            thickness=20,
            line=dict(color="black", width=0.5),
            label=list(label_map.keys())
        ),
        link=dict(
            source=links_df['source'],
            target=links_df['target'],
            value=links_df['count']
        ))])
    sankey.update_layout(title_text=f"Sankey Diagram - 실험군: {group}", font_size=10)
    sankey.show()
````

### 통계적 유의성 검정
- 각 특성별 결과값이 통계적으로 유의한 결과값인지 검증하겠습니다.
- 두 데이터는 독립표본이므로 독립표본검정을 수행하여 통계적 유의성 검정을 진행합니다.
- 두 그룹간 ABT 성과의 분포 차이가 유의미한지 분석을 위해 카이제곱 분포 검정을 통해 통계적 유의성 검정을 진행합니다.

````python
import pandas as pd
from scipy.stats import ttest_ind, chi2_contingency

# 비교 대상 KPI 목록
metrics = ['click_service', 'click_count', 'page_view_count', 'session_duration_sec', 'dropout_flag']

# 결과 저장용 딕셔너리
results = []

# 그룹 분리
group_H0 = df[df['experiment_group'] == 'H0']
group_H1 = df[df['experiment_group'] == 'H1']

# 평균 비교 (T-test for continuous variables)
for metric in metrics:
    stat, pvalue = ttest_ind(group_H0[metric], group_H1[metric], equal_var=False)
    results.append({
        "metric": metric,
        "H0_mean": group_H0[metric].mean(),
        "H1_mean": group_H1[metric].mean(),
        "t_statistic": stat,
        "p_value": pvalue,
        "test_type": "t-test"
    })

# 이항 변수 chi-square test (click_service, dropout_flag)
for metric in ['click_service', 'dropout_flag']:
    contingency = pd.crosstab(df['experiment_group'], df[metric])
    chi2, pval, _, _ = chi2_contingency(contingency)
    results.append({
        "metric": metric + " (chi2)",
        "H0_mean": group_H0[metric].mean(),
        "H1_mean": group_H1[metric].mean(),
        "chi2_statistic": chi2,
        "p_value": pval,
        "test_type": "chi-square"
    })

# 결과를 DataFrame으로 정리
results_df = pd.DataFrame(results)

# 전체 출력 형식을 소수점 6자리 실수로 변경
pd.set_option('display.float_format', '{:.6f}'.format)

results_df

# comment
# AB 테스트 결과, 실험군(H1)의 page_view_count 지표에서 대조군(H0) 대비 통계적으로 유의한 차이가 관측되었습니다 (p < 0.001).
# 이는 실험 그룹의 평균 페이지뷰가 약 1.35회 이상 감소한 것을 의미합니다.
# 다만, 단순한 페이지 뷰 감소는 UX 효율 개선으로도 해석 가능하며, 동시에 관심 저하나 이탈 증가로도 해석될 수 있습니다.
# 따라서 click_service, conversion, dropout_flag 등의 보완 지표를 함께 분석하여
# 본 결과가 실질적인 서비스 가치 향상으로 이어졌는지를 종합적으로 평가하는 것이 바람직합니다.
````
