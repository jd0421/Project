# [1] 개요

1. 배경

```
- 앱의 홈 화면 상단에는 모바일 기준 총 23개의 배너가 롤링 방식으로 노출됩니다.
- 현재 배너 중 일부는 쿠폰 지급을 통해 높은 전환율을 유도하고 있습니다.
- 다만 전환 이후에도 해당 동일한 배너가 반복 노출되는 경우보다, 후단의 미전환 배너의 노출을 통해 광고 피로도 감소 및 전체 배너 전환율 개선이 가능하지 않을까를 고민하며 ABT 실험 설계를 시행하게 되었습니다.
- 특히 유료 광고 배너의 경우 고정 단가로 운영될 것으로 예상되며, 효율적 노출 전략이 광고주 성과 및 당사의 내부 수익성에 직접적인 영향을 미치게 됩니다.
- 이에 따라, 전환 이후 동일 배너 대신 후속 배너로 자연스럽게 전환하는 방식이 전환율 개선에 기여하는지 검증하고자 합니다.
```

2. 실험 개요

```
- 전환 전/후 배너 1의 노출 여부가 사용자 행동 및 매출에 미치는 영향을 측정하기 위한 목적 아래, A/B 테스트 형식으로 설계·진행되었습니다.
    - 그룹 A : 전환 후 동일 배너 노출 그룹
    - 그룹 B : 전환 후 해당 배너 제외 노출 그룹
    - 귀무가설(H₀): A, B 그룹 간 전환율에는 유의미한 차이가 없다.
    - 대립가설(H₁): B 그룹의 전환율이 A 그룹보다 유의미하게 높다.

- 이를 기반으로 주요 KPI 및 가드레일 지표에 대한 정량 분석, 통계적 유의성 검정, 효과 크기 분석, 기여도 평가 등 분석 수행하였습니다.
```

3. 주요 결과 요약

```
1) KPI 성과 요약
그룹 B에서 구매 수는 +6.8%, 매출은 +1.3% 증가
→ 반면, 구매당 매출은 -5.2% 하락

이는 더 많은 사용자가 구매했지만, 평균 객단가는 소폭 하락한 것으로 해석됩니다.

2) 통계적 유의성 및 효과 크기
주요 KPI의 p-value는 모두 유의 수준(0.05) 을 넘지 않아,
→ 통계적으로 일반화할 수 있는 효과는 미약한 수준

Cohen’s d 효과 크기 분석 결과 역시 대부분 ‘무시할 수준’ 또는 ‘작은 효과’로 분류됨
→ 수치상 변화는 존재하나, 실질적 사용자 행동 변화는 미미한 수준으로 해석됩니다.

3) 가드레일 지표
Revenue per Impression (RPI): 그룹 B가 A 대비 +1.65% 향상
→ 손익 측면에서 비열등성 기준(NIM = -1%)을 통과, 실험으로 인한 매출 손실은 없음

타 배너 CTR 안정성: A/B 간 차이 +0.03%p로 매우 미미
→ 배너 1 조작이 전체 UX에 부정적 영향을 미치지 않음이 입증됨

4) 기여도 평가 및 증분 효과
Multi-Touch 기여도 분석 결과, 구매자 중 1.3%가 복수 배너 클릭 이력을 보유
→ 다소 미미하지만, 일부 사용자에겐 간접적인 광고 기여 가능성 시사

Incremental Revenue 분석에서는 그룹 A가 +56.6%의 매출 증대 효과를 보여줌
→ p-value = 0.00009로 통계적으로도 유의하며, 배너 노출 자체가 구매의사 또는 브랜드 자극에 긍정적 영향을 미쳤을 가능성
```

4. 결론 및 제안

```
- 성과 향상은 일부 존재하나, 전체 KPI 효과는 제한적
    → 특히 객단가 하락은 보완이 필요한 부분이며, 매출 확대의 지속성은 미확정적

- 현재 구조는 ‘방어적인 실험’으로 평가 가능
    - RPI 유지 + 타 배너 영향 없음

    - 단일 클릭 중심 분석에서 벗어나, 노출·전환 간의 다채널 기여도 모델링 필요
    - 초기 결과를 바탕으로, 크로스셀 배너 전략 또는 UX-매출 Trade-off 실험군 설계에 적용 가능
```

| 항목                | A 그룹    | B 그룹 |

```
    - 0.2 = small, 0.5 = medium, 0.8 = large
- 효과 크기(Cohen’s d): 0.143
- 분석 결과:
    - 각 그룹당 최소 샘플 수 ≈ 2,500명
```

### [4-1-1] 정량 지표 비교

```   
- Comment
	- 구매 수 +6.8%, 매출 +1.3% 증가하며 매출액 대비 구매 수가 높게 증가하였고, 상대적으로 구매당 매출은 소폭 하락 (-5.2%)하였습니다.
	- 실험 그룹의 구매 수 증대를 정량적으로 확인되었기에, 고매출 소재의 배너 광고 게시 등을 통해 구매 당 매출 개선을 도모해볼 수 있겠습니다.
```


````python
# 그룹별 purchase rate 및 revenue per impression 확인
summary_df = df.groupby("group").agg(
    total_users=("user_id", "nunique"),
    total_impressions=("impression", "sum"),
    total_purchases=("purchase", "sum"),
    total_revenue=("revenue", "sum")
).reset_index()

# KPI 계산
summary_df["purchase_rate"] = summary_df["total_purchases"] / summary_df["total_impressions"]
summary_df["revenue_per_impression"] = summary_df["total_revenue"] / summary_df["total_impressions"]
summary_df["revenue_per_purchase"] = summary_df["total_revenue"] / summary_df["total_purchases"]

# 그룹별 KPI 정량 비교 요약 결과에서 B - A 차이(diff) 계산
summary_df = summary_df.set_index('group')

# diff 행 추가 (B - A)
summary_df.loc['diff (B - A)'] = summary_df.loc['B'] - summary_df.loc['A']

# 그룹 A와 B를 행이 아닌 열 기준으로 정리하여 비교 테이블 생성
summary_transposed = summary_df.T.copy()

# A, B 값으로부터 diff와 증감율 계산
summary_transposed['diff (B - A)'] = summary_transposed['B'] - summary_transposed['A']
summary_transposed['diff (%)'] = ((summary_transposed['B'] - summary_transposed['A']) / summary_transposed['A']) * 100

summary_transposed.style.background_gradient(cmap='coolwarm', subset = ['diff (%)'])
````

### [4-1-2] 통계적 유의성 검정

```
- Comment
	- KPI 지표 모두 p-value가 유의 수준(0.05) 이상으로 나타나, 실험군의 퍼포먼스 변화는 관측된 수치 차이는 존재하나, 통계적으로 일반화하기엔 부족한 근거로 해석됩니다.
	- 다만, 전환율 및 매출 지표가 실질적으로 증가한 경향은 확인되므로, 효과 크기(Cohen's d) 분석 등을 병행하여 비즈니스적 의미를 보완하고자 합니다.
```


| KPI                    | Cohen’s d | 효과 크기 판단           | 해석 요약                        |
| ---------------------- | --------- | ------------------ | ---------------------------- |
| CTR                    | -0.0158   | 무시 가능 (Negligible) | 그룹 간 클릭률 차이는 사실상 없음          |
| Purchase Rate          | -0.0071   | 무시 가능 (Negligible) | 전환율 차이도 거의 없음                |
| Revenue per Impression | -0.0015   | 무시 가능 (Negligible) | 노출 대비 매출 효율 변화도 없음           |
| Revenue per Purchase   | 0.1464    | 작음 (Small)         | 실험군(B)의 구매당 매출이 다소 낮은 경향을 보임 |


```
- Comment
    - Cohen's d 검정 결과 전반적으로 행동상의 효과 크기는 미미하거나 없으며, 실질적 영향력은 낮다고 평가됩니다.
```

````python
from scipy.stats import ttest_ind

# 구매율, 구매수, 매출, 구매당 매출 항목으로 집계된 KPI 재사용
summary_df

# 통계 검정을 위해 원시 데이터를 다시 그룹핑
grouped = df.groupby('group')

# 각 그룹의 구매율 = purchase / click, 구매당 매출 = revenue / purchase
group_a = grouped.get_group('A')
group_b = grouped.get_group('B')

# 유의성 검정: Purchase Rate (구매율)
purchase_rate_a = group_a['purchase']
purchase_rate_b = group_b['purchase']
t_stat_purchase, p_val_purchase = ttest_ind(purchase_rate_a, purchase_rate_b)

# 유의성 검정: Revenue (매출)
revenue_a = group_a['revenue']
revenue_b = group_b['revenue']
t_stat_revenue, p_val_revenue = ttest_ind(revenue_a, revenue_b)

# 유의성 검정: Revenue per Purchase (구매당 매출)
# purchase == 0 인 경우 제외
valid_a = group_a[group_a['purchase'] > 0]['revenue'] / group_a[group_a['purchase'] > 0]['purchase']
valid_b = group_b[group_b['purchase'] > 0]['revenue'] / group_b[group_b['purchase'] > 0]['purchase']
t_stat_rpp, p_val_rpp = ttest_ind(valid_a, valid_b)

# 결과 정리
import pandas as pd

stats_df = pd.DataFrame({
    'KPI': ['Purchase Rate', 'Revenue', 'Revenue per Purchase'],
    't-statistic': [t_stat_purchase, t_stat_revenue, t_stat_rpp],
    'p-value': [p_val_purchase, p_val_revenue, p_val_rpp]
})

import numpy as np
stats_df['Significant (p<0.05)'] = np.where(stats_df['p-value'] < 0.05, 'Yes', 'No')

stats_df
````

### [4-2-2] 유의성 검정 (비열등성 등)

| 결론          | 비열등성 입증 실패       |

- RPI는 실험군(B)이 대조군(A)보다 평균적으로 높지만,
설정한 열등성 한계(NIM = -25)를 기반으로 했을 때,
이 차이가 통계적으로 유의미하게 비열등하다고 입증되지는 못함.

---
2. Other Banners CTR Stability – 정량 비교 결과

- 실험은 banner 1에 대한 노출 변경을 테스트한 것이며,
- 나머지 2~23번 배너들의 평균 CTR에 대한 영향을 살펴보는 것이 핵심입니다.

| 항목                   | 값                 |
| -------------------- | ----------------- |
| A 그룹 (2\~23번 평균 CTR) | 0.0578            |
| B 그룹 (2\~23번 평균 CTR) | 0.0575            |
| 차이 (B - A)           | -0.0003 (-0.03%p) |
| 기준 threshold         | ±0.005 (±0.5%p)   |
| 결론               | 영향 없음         |

- 기타 배너 CTR에는 유의미한 영향이 없음
- → 실험 변경이 다른 콘텐츠 영역을 침해하거나 사용자 관심을 뺏지 않음.

- Comment :

    - RPI 검정 결과 : 실험군이 평균적으로 RPI는 높지만, 명확하게 우수하다고 입증되지는 못함. 보수적인 관점에서는 유지 결정 보류, 그러나 긍정적 방향성이 있으므로 추가 실험 또는 장기 추적 권장

    - Other CTR 안정성 : 실험이 타 배너의 성과를 침해하지 않음이 입증됨. 서비스 UX 전체에 부정 영향 없다는 근거 확보, 채널 전반의 안정성 확보

````python
import scipy.stats as stats
import numpy as np

# 가드레일 지표: 1. Other Banners CTR Stability (banner_id ≠ 1)
ctr_A = df[(df['group'] == 'A') & (df['banner_id'] != 1)].groupby('user_id').agg({
    'click': 'sum',
    'impression': 'sum'
})
ctr_A['ctr'] = ctr_A['click'] / ctr_A['impression']
ctr_B = df[(df['group'] == 'B') & (df['banner_id'] != 1)].groupby('user_id').agg({
    'click': 'sum',
    'impression': 'sum'
})
ctr_B['ctr'] = ctr_B['click'] / ctr_B['impression']

# 독립표본 t검정 (다른 배너 CTR)
t_ctr, p_ctr = stats.ttest_ind(ctr_A['ctr'].fillna(0), ctr_B['ctr'].fillna(0), equal_var=False)

# 가드레일 지표: 2. Revenue per Impression (RPI)
rpi_A = group_A['revenue'] / group_A['impression']
rpi_B = group_B['revenue'] / group_B['impression']

# 비열등성 검정을 위해 NIM (Non-Inferiority Margin) 설정
NIM = -25  # 예시: 25원 이하 하락은 비열등한 것으로 간주하지 않음

# 비열등성 검정: 그룹 B - 그룹 A >= -NIM 를 증명해야 함
diff_mean = np.mean(rpi_B.fillna(0)) - np.mean(rpi_A.fillna(0))
std_pooled = np.sqrt(
    (np.var(rpi_A, ddof=1) / len(rpi_A)) +
    (np.var(rpi_B, ddof=1) / len(rpi_B))
)
t_noninf = (diff_mean + NIM) / std_pooled  # +NIM because we test B is not worse than A by more than NIM
p_noninf = 1 - stats.norm.cdf(t_noninf)  # one-sided test

# 결과 정리
guardrail_test_result = pd.DataFrame({
    'Metric': ['Other Banners CTR Stability', 'Revenue per Impression (Non-Inferiority)'],
    'Test Type': ['Two-sided t-test', 'One-sided non-inferiority test'],
    'Test Statistic': [t_ctr, t_noninf],
    'p-value': [p_ctr, p_noninf],
    'Interpretation': [
        '차이 없음' if p_ctr > 0.05 else '차이 있음',
        '비열등성 입증' if p_noninf < 0.05 else '비열등성 입증 실패'
    ]
})

guardrail_test_result
````

### [4-3-1] Multi-Touch 분석

- Objective :
    - 사용자가 여러 배너를 클릭한 후 구매한 경우,
최종 클릭 배너 외에 앞선 클릭 배너들의 간접 기여도(Assisted Conversion)를 정량적으로 평가합니다.
    - 단일 배너 전환에 집중된 판단을 보완하고, 전체 캠페인의 협업 효과를 측정합니다.

- 가정 :
    - 동일 user_id가 여러 row에 걸쳐 노출 → 클릭 → 전환한 기록이 존재
    - banner_id가 1인 배너의 클릭 이후 구매는 라스트터치(Last Touch)
    - 2~23번 배너 클릭 → 이후 banner_id=1 클릭 후 구매는 Assisted Touch

- How to :

| 단계 | 설명                                        |
| -- | ----------------------------------------- |
| ①  | user\_id 단위로 클릭 이력 정렬 (시간 순)              |
| ②  | 구매자 중 banner 1을 클릭한 유저를 추출                |
| ③  | 해당 유저가 banner 1 클릭 이전에 클릭한 다른 배너가 있었는지 확인 |
| ④  | 이 경우를 Assisted Conversion으로 집계        |

- Results :

| - | 항목 | 값 |
| -- | -- | -- |
|0	|총 구매자 수	|381.0000
|1	|Assisted Conversion 수	|5.0000
|2	|Assisted Conversion 비율	|0.0131

- Summary
    - 전체 구매자 중 1.31%의 유저가 최종 클릭 이전에도 다른 배너를 클릭한 기록이 있습니다.
    - 이는 복수의 배너 노출 및 클릭이 구매 행동에 간접적으로 영향을 줄 수 있음을 시사합니다.
    - 다만 수치가 낮기 때문에, 현재 구조나 기간에서는 Multi-Touch 전환 기여가 미미한 수준입니다.

- 비즈니스 인사이트 제안
    - 향후 더 정교한 클릭 로그 수집 체계나, longer-term attribution window 확보가 가능하다면, 배너 간 협업 전환 효과(Multi-Touch Contribution) 를 더 정확히 측정할 수 있을 것입니다.
    - 해당 유저군을 중심으로 "유입 루트 다각화"나 “리타겟팅 강화” 전략을 설계해볼 수 있습니다.

````python
# 1. 구매한 유저 목록 추출
purchased_users = df[df['purchase'] > 0]['user_id'].unique()

# 2. 각 유저의 클릭 배너 히스토리 수집
click_logs = df[(df['user_id'].isin(purchased_users)) & (df['click'] > 0)]
click_history_by_user = click_logs.groupby('user_id')['banner_id'].apply(list)

# 3. 각 유저의 구매 시점의 클릭 배너 추출
purchase_logs = df[(df['user_id'].isin(purchased_users)) & (df['purchase'] > 0)]
last_click_banner_by_user = purchase_logs.groupby('user_id')['banner_id'].last()

# 4. 어시스트(선행 클릭) 여부 판단
assisted_conversion_users = []

for user in purchased_users:
    history = click_history_by_user.get(user, [])
    last_click = last_click_banner_by_user.get(user)
    if last_click in history:
        history_without_last = [b for b in history if b != last_click]
        if len(history_without_last) > 0:
            assisted_conversion_users.append(user)

# 결과 요약
total_buyers = len(purchased_users)
assisted_buyers = len(assisted_conversion_users)
assisted_rate = assisted_buyers / total_buyers if total_buyers > 0 else 0

# total_buyers, assisted_buyers, round(assisted_rate, 4)

result = pd.DataFrame({
    '항목': ['총 구매자 수', 'Assisted Conversion 수', 'Assisted Conversion 비율'],
    '값': [total_buyers, assisted_buyers, round(assisted_rate, 4)]
})

result

````

### [4-3-2] Incremental Revenue 분석

- Objective :
    - Incremental Revenue Attribution은 특정 요소(여기서는 배너 1의 노출 여부)가 실제 매출 증가에 미친 영향을 측정하는 분석 기법입니다.
    - 실험군(배너 1 노출)과 통제군(비노출) 간의 평균 Revenue 차이를 비교하고, 이 차이가 통계적으로 유의미한지를 판단합니다.

- Summary :
    - 실험군 그룹이 대조군 그룹보다 약 56.6% 높은 수익을 기록했습니다.
    - t-검정 결과 p-value = 0.00009로, 통계적으로 유의미한 차이가 존재합니다 (유의수준 0.05 기준).
    - 이는 배너 1 노출이 실제로 매출 증대에 기여했을 가능성을 시사합니다.

-  비즈니스 관점 해석 :
    - 배너 1 노출 시 사용자당 매출이 유의하게 상승했습니다.
        -  단순 노출 자체가 구매로 이어지지 않더라도 고객의 구매의도 또는 브랜드 인지도를 자극했을 가능성이 높습니다.
    - 특히, 단일 클릭 중심의 Last-touch attribution이 아닌, 노출 기반의 기여도 관점에서 의미 있는 인사이트로 볼 수 있습니다.

````python
import pandas as pd
import scipy.stats as stats

# banner_id == 1 노출 여부 파악 (impression > 0 인 경우 노출로 간주)
banner_1_exposed_users = df[(df['banner_id'] == 1) & (df['impression'] > 0)]['user_id'].unique()
df['exposed_to_banner_1'] = df['user_id'].isin(banner_1_exposed_users)

# 유저 단위로 revenue 집계
user_revenue = df.groupby(['user_id', 'exposed_to_banner_1'])['revenue'].sum().reset_index()

# 그룹 분리
exposed = user_revenue[user_revenue['exposed_to_banner_1'] == True]['revenue']
not_exposed = user_revenue[user_revenue['exposed_to_banner_1'] == False]['revenue']

# 각 그룹 평균 수익
mean_exposed = exposed.mean()
mean_not_exposed = not_exposed.mean()

# 수치 차이와 비율 차이
revenue_diff = mean_exposed - mean_not_exposed
revenue_diff_pct = (revenue_diff / mean_not_exposed) * 100 if mean_not_exposed != 0 else None

# 독립표본 t-검정 수행
t_stat, p_value = stats.ttest_ind(exposed, not_exposed, equal_var=False)

# 결과 요약
summary = {
    '평균 Revenue (노출)': mean_exposed,
    '평균 Revenue (비노출)': mean_not_exposed,
    'Revenue 차이': revenue_diff,
    'Revenue 증감율 (%)': revenue_diff_pct,
    't-통계량': t_stat,
    'p-값': p_value
}

summary_df = pd.DataFrame([summary])
summary_df
````

###  [5-2] 주요 KPI 정의 및 계산 방식

| KPI | 정의 | 계산식 |
|-----|------|--------|

### [5-3] 검정 방식 및 해석 기준

#### 1) 통계적 유의성 검정 (p-value 기준)

- 사용 검정: 독립표본 t-검정 (`scipy.stats.ttest_ind`)  
- 유의수준 (α): 0.05  
- 해석 기준:
  - `p < 0.05`: 유의미한 차이 있음  
  - `p ≥ 0.05`: 유의미한 차이 없음 (귀무가설 기각 불가)

#### 2) 효과 크기 분석 (Cohen’s d)

- 사용 방식: 표준화된 평균 차이  
- 공식:

$$
d = \frac{\mu_1 - \mu_2}{\sigma_\text{pooled}} \quad \text{where }
\sigma_\text{pooled} = \sqrt{ \frac{(n_1 - 1)s_1^2 + (n_2 - 1)s_2^2}{n_1 + n_2 - 2} }
$$

- 해석 기준 (Cohen, 1988):

| 절댓값 d | 해석 |
|----------|------|
| 0.0 ~ 0.2 | 없음 또는 무시 가능 (Negligible) |
| 0.2 ~ 0.5 | 작은 효과 (Small) |
| 0.5 ~ 0.8 | 중간 효과 (Medium) |
| 0.8 이상 | 큰 효과 (Large) |


### [5-4] 비열등성 검정 (Non-Inferiority Test)

- 사용 목적: 실험군(B)의 KPI가 대조군(A)보다 유의미하게 열등하지 않음을 입증  
- 귀무가설 (H₀): B 그룹의 성과가 A 그룹보다 NIM (Non-Inferiority Margin) 이상 열등하다  
- 대립가설 (H₁): B 그룹의 성과는 A보다 NIM 이상 열등하지 않다

- 예시: RPI 기준 비열등성 검정
  - NIM (비열등성 한계) = -1%
  - Δ (A - B의 평균 차이) = -1.65%
  - → Δ < NIM 이므로, 귀무가설 기각 불가 → 비열등성 통과

### [5-5] 기여도 분석 (Attribution Analysis)

#### 1) Assisted Conversion 정의

- Assisted Conversion (간접 기여 전환):  
  사용자가 최종 전환(구매) 이전에 여러 개의 배너를 클릭한 경우, 마지막 클릭 외의 모든 클릭들을 '간접 기여'로 간주함.
  
- 적용 가정:
  - 동일한 `user_id`가 여러 row에 등장
  - 전환은 오직 한 번이며, 최종 클릭 배너는 `Last-Touch`, 이전 클릭들은 `Assisted Touch`
  
- 지표 해석:
  - `Assisted Conversion 비율 = 간접 기여 전환자 수 / 전체 전환자 수`
  - 값이 높을수록 다채널 / 다배너의 연속 노출 효과가 존재할 수 있음

#### 2) Incremental Revenue Attribution 정의

- Incremental Revenue Attribution (증분 매출 기여도):  
  특정 요소(예: 배너 1 노출)의 매출 기여도를 정량화  
  → 실험군 vs 통제군 간 평균 Revenue 차이 비교

- 사용 방식:  
  - 그룹 A, B의 `revenue` 평균 비교  
  - 독립표본 t-검정을 통해 평균 차이의 통계적 유의성 판단

- 적용 가정:  
  - 두 그룹은 동질성을 갖고 있으며, 배너 노출 외 다른 조건은 동일  

- 지표 해석:
  - `Incremental Revenue = mean(B) - mean(A)`
  - `p-value < 0.05`: 배너 노출이 통계적으로 유의미한 매출 증가를 가져왔다고 판단 가능
